A variable is a named storage location used to store data in a program.

```python
# declaring and assigning variables

age = 21
height = 6.1
name = "max"
is_student = True


## printing the variables

print("age :", age )
print("height :", height )
print("name :", name )
print("is_student :", is_student )
```
