A data type tells Python what kind of data a variable stores.

```python
## integers example

age = 35
type(age)
```

```python
## floating point datatype

height = 5.11
print(height)
print(type(height))
```

```python
## string datatype example

name = "situ"
print(name)
print(type(name))
```

```python

## boolean datatype
is_true = True
type(is_true)
```
