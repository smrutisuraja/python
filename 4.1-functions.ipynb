A funnction is a block of code that perform a specific task. function helps in organizing code , reusing the code and improving redability.

```python
## syntax
def function_name(parameters):
    ''' Docstring'''
    ## function body
    return expression
```

```python
 ## why function
num = 24
if num %2 == 0:
    print("the number is even")
else:
    print("the number is odd")
```

```python
## creating function
def even_or_odd(num):
    '''this function finds even or odd'''
    if num %2 == 0:
        print("the number is even")
    else:
        print("the number is odd")
        
```

```python
## call the function
even_or_odd(24)
```

```python
## function with multiple parameters

def add(a , b):
    return a+b

result = add(2,4)
print(result)
```

```python
## default parameters

def greet(name ="guest" ):
    print(F"hello {name} wellcome to the course")

greet("situ")
```

```python
## variable length arguments
## positional and key word arguments

def print_numbers(*situ):
    for number in situ:
        print(number)
```

```python
print_numbers (1,2,3,4,5,6,7,8,"situ")
```

```python
## positional arguments
def print_numbers(*args):
    for number in args:
        print(number)
```

```python
print_numbers = (1,2,3,4,5,6,7,8,"situ")
```

```python
## keyword arguments (all the parameters in key value pairs)

def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name = "situ",age = 21, country = "india")
```

```python
def print_details(*args,**kwargs):
    for val in args:
        print(f"positional argument:{val}")
        
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details  (1,2,3,4,5,"situ",name = "situ",age = 21, country = "india")
```
