# map() function in python

the map()function applies a given function to all items in an input list(or any other iterable) and return a map oject(an iterator). this is particularly useful for transforming data in alist comprehensively.

```python
def square(x):
    return x*x

square(45)
```

```python
numbers = [1,2,3,4,5,6,7,8]

list(map(square, numbers))
```

```python
## lambda function with map()
numbers = [1,2,3,4,5,6,7,8]

list(map(lambda x:x**2, numbers))
```

```python
## can be map mutiple iterables
# i want to add 5,7,9
numbers1 = [1,2,3]
numbers2 = [4,5,6]

added_numbers = list(map(lambda x,y:x+y,numbers1,numbers2 ))
print(added_numbers)
```

```python
## map() to convert list of string to integer

str_numbers = ["1" , "2" , "3" , "4" , "5" ]
int_numbers = list(map(int,str_numbers))
print(int_numbers)
```

```python
# convert lower word to upper word using map case
words = ["apple","banana","cherry"]
upper_word = list(map(str.upper,words))
print(upper_word)
```

```python
def get_name(person):
    return person["name"]

people =[
    {"name" : "situ", "age" : 21},
    {"name" : "sipu", "age" : 15},
]
list(map(get_name,people))
```
