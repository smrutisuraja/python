# CREATE A TO DO LIST TO KEEP TRACK OF TASKS

```python
to_do_list = ["complete seminar report" , "get some notes", "study"]

## adding some tasks
to_do_list.append("playing")
to_do_list.append("complete project")

# removing a completed tasks
to_do_list.remove( "get some notes")

# checking if a task is in the list
if "complete seminar report" in to_do_list:
    print("do not forget to complete seminar report")

print("to do list remaining")
for task in to_do_list:
    print(f"{task}")
```

# ORGANIZING STUDENT GRADES

* create a list to store and calculate average grades for students

```python
## organizing students grade
grades = [85,92,78,90,88]

# adding a new grade
grades.append(95)

#calculating the average grade
average_grades = sum(grades)/len(grades)
print(f"average_grades: {average_grades:.2f}")

# finding the highest and lowest grades

highest_grades = max(grades)
lowest_grades = min(grades)
print(f"Highest grade :{highest_grades}")
print(f"lowest grade :{lowest_grades}")
```

# MANAGING AN INVENTORY


* use a list to manage inventory items in a store

```python
## managing an inventory
inventory = ["apples", "oranges", "banana", "grapes"]

## add new item
inventory.append("graves")

## removing an item tha is out of stock
inventory.remove("banana")

#checking of an item in the stock
item = "orange"






if item in inventory:
    print(f"{item} are in stock")
else:
    print(f"{item} are out of stock")

#printing the inventory

print("inventory list:")
for item in inventory:
    print(f"- {item}")
```

# COLLECTING A USER FEEDBACK
* use a lisst to collect and analyze user feedback

```python
## collecting user feedback
user_feedback = ["graet service","very satisfying","could be better", "excellent experience"]
## add new feedback
user_feedback.append("not happy with this service")

## counting specific feedback
positive_feedback_count = sum(1 for comment in user_feedback if "graet" in comment.lower() or "excellent" in comment.lower())
print(f"positive feedback count: {positive_feedback_count}")

#printing all feedback

print("user_feedback")
for comment in user_feedback:
    print(f"-{comment}")
```
