```python
## read a whole file

with open('example.txt','r') as file:
    content = file.read()
    print(content)
```

```python
## read a file line by line
with open('example.txt','r') as file:
    for line in file:
        print(line.strip())## .strip() removes new line charecter
```

```python
# writing a file(Overfitting)
with open('example.txt','w') as file:
    file.write('Hello world!\n')
    file.write('this is a new line')
```

```python
# write a file without overwritting
with open('example.txt','a') as file:
    file.write("append operation taking place!\n")
```

```python
# writting list of lines to file

lines = ['first line\n','second line\n','third line\n']
with open('example.txt','a') as file:
    file.writelines(lines)
```

```python
# binary files
# writing the binary file

data = b'\x00\x01\x02\x03\x04'
with open('example.bin', 'wb') as file:
    file.write(data)
```

```python
# reading binary file
with open('example.bin','rb') as file:
    content = file.read()
    print(content)
```

```python
## read the ccontent from source text file and write in the destination text file
# copying a text file

with open('example.txt','r') as source_file:
    content = source_file.read()

with open('destination.txt','w') as destination_file:
    destination_file.write(content)
```

```python
# read a txt file and count the nuber of lines ,words and charecters
# Counting lines, words, and characters in a text file
def count_text_file(file_path):
    with open(file_path, 'r') as file:
        lines = file.readlines()
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
    return line_count, word_count, char_count
file_path = 'example.txt'
lines, words, characters = count_text_file(file_path)
print(f'Lines: {lines}, Words: {words}, Characters: {characters}')
```

the w+ mode in python is used to open a file for both reading and writing. if the file does not exist, it will be created , if the file exists, its contents is truncated(i.e., the file is overwritten)

```python
# writing and then read a file

with open('example.txt','w+') as file:
    file.write("hello world!\n")
    file.write("This is a new line\n")

    ## move the cursor to beginning
    file.seek(0)

    # read tyhe content of the file
    content = file.read()
    print(content)
```
