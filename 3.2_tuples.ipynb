```python
## paacking and unpacking tupple
# packing
packed_tuple = 1 ,"sneha", 3.14
print(packed_tuple)
```

```python
# unpacking a tuple
a,b,c = packed_tuple
print(a)
print(b)
print(c)
```

```python
# unpacking with *
number = (1,2,3,4,5,6,7)
first,*middle, last = number
print(first)
print(middle)
print(last)
```

```python
nested_tuple = ((1,2,3),("a","b","c"),(True ,False))

#access the element inside a tuple
print(nested_tuple[0])
print(nested_tuple[1][2])
print(nested_tuple[2][1])
```

```python
# itearting over nested tuples
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end = " "  )

    print()
```
