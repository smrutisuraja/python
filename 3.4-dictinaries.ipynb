## Dictionaries are unordered collection of items. they stores data in key value pairs . key must be unique and immutable(e.g., strings , numbers , tuples), while values can be of any type.

```python
## creating dictionaries
empty_dict = {}
print(type(empty_dict))
```

```python
empty_dict = dict()
empty_dict
```

```python
student = {"name":"krish" , "age" : 21, "grade":24}
print(student)
print(type(student))
```

```python
## single key is always used
student = {"name":"krish" , "age" : 21, "name" : "situ" }
print(student)
```

```python
## accessing dictionary elements
student = {"name":"krish" , "age" : 21, "grade":"A"}
print(student)
```

```python
## accesing dictionary element
print(student["grade"])
print(student["age"])

## Accessing using get() method
print(student.get("grade"))
print(student.get("last_name"))
print(student.get("last_name" , "not avilable"))
```

```python
## modifying dictionary elements
## dictionary are mutable , so you can add , update or delete elements
print(student)
```

```python
student["age"] = 33  ## update value for the key
print(student)
student["address"] = "india" ## added a new key and value
print(student)
```

```python
del student["grade"]  # delete key and value pair
print(student)
```

```python
## dictionary methods

keys = student.keys()  ## get all the keys
print(keys)
values = student.values() ## get all values
print(values)

items = student.items() ## get all key pairs
print(items)
```

```python
student_copy = student
print(student)
print(student_copy)
```

```python
student_key = student
print(student)
print(student_key)
```

```python
student["name"] = "situ"
print(student)
print(student_copy)
```

```python
student_copy1 = student.copy() ## it is responsible for shallow copy
print(student)
print(student_copy1)
```

```python
student["name"] = "situ1"
print(student)
print(student_copy)
print(student_copy1)
```

```python
## iterating over dictionaries
## you can use loops to iterarte over dictionary ,keys , values or items

# iterating over keys
for keys in student.keys():
    print(keys)
```

```python
# iterating over values
for values in student.values():
    print(values)
```

```python
## iterate over key value pairs
for keys,values in student.items():
    print(f"{keys}:{values}")
```

```python
## nested dictionary
students  = {
    "student1" : {"name":"situ","age":21},
    "student2" : {"name":"sipu","age":15}

}
print(students)
```

```python
## access nested dictionary elements
print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
students.items()
```

```python
## iterating over nested dictinaries
for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
## dictionary comprehensin
squares = {x : x**2 for x in range(5)}
print(squares)
```

```python
square = {}
for x in range(5):
    square [x] = x ** 2
print(square)
```

```python
## conditional dictionary comprehension
even = {x:x ** 2 for x in range(10) if x %2 == 0}
print(even)
```

```python
## practical examples


## use a dictionary to count the frequency of element in list

numbers = [1,2,2,3,3,3,4,4,4,4] 

frequency = {}

for number in numbers:
    if number in frequency:
        frequency [number]+=1
    else:
        frequency[number] = 1

print(frequency)
```

```python
## merge 2 dictionaries into one 

dict1 = {"a":1,"b":2}
dict2 = {"b":3,"c":4}
merged_dict={**dict1,**dict2}
print(merged_dict)
```
