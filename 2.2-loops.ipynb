```python
# while loop

## the while loop continues to execute as long as the condition is true,

count = 0

while count<5:
    print(count)
    count= count+1
```

```python
## loop control statements

## break
## the break statements exists the loop permatuely

# break statements

for i in range(10):
    if i == 5:
        break
    print(i)
# in this condition when we put break then the loop is end
```

```python
## continue

## the continue statement skip the current iteration and continues with the next

for i in range(10):
    if i%2 == 0:
        continue
    print(i)
```

```python
for i in range(10):
    if i == 4:
        continue
    print(i)
```

```python
# pass

## the pass statements is a null operation and; it does nothing.

for i in range(5):
    if i == 3:
        pass
    print(i) 
```

```python
for i in range(5):
    if i == 3:
        print("the number is",i)
        pass
    print(i)
```

```python
## nested loops

## loop inside a loop
for i in range(3):
    for j in range(2):
        print(f"i: {i} and j : {j}")
# in this case first i execute 1 time then it go to j then j is execute full then it go to i and it is continue in a loop
```

```python
## examples - calculate the sum of first n natural number using a while and for loop

## while loop

n =10
sum  = 0
count = 1

while count <= n:
    sum = sum + count
    count = count+1

print("sum of the first ten natural number is :", sum)
```

```python
## for loop

n =10
sum  = 0

for i  in range(11):
    sum = sum + i
print(sum) 
```

```python
## examples - prime numbers between 1 and 100


for num in range(1, 101):
    if num>1:
        for i in range(2, num):
            if num%i == 0:
                break

        else:
            print(num)
```
