```python
# nested conditional statements
# you can place one or more if , elif, or else statements inside another if ,elif, or else statements to createnested conditional statements
# the number is positive ,negative, even ,odd
num = int(input("Enter the number"))

if num > 0:
    print("the nmber is positive")
    if num%2 == 0:
        print("the number is even")
    else:
        print("the number is odd")
else:
    print("the number is zero or negative")    
```

```python
## assignment
## simple calculator program
num1 = float(input("enter first number: "))
num2 = float(input("enter second number: "))
operation = input("enter operation(+,-,*,/)")

if operation == "+":
    result = num1 + num2
elif operation == "-":
    result = num1 - num2
elif operation == "*":
    result = num1 * num2
elif operation == "/":
    if num2 != 0:
        result = num1 / num2
    else:
        result = "error ! division by zero."
else:
    result = "invalid operation"

print("result:", result)    
```

```python
## determine the cricket game stadium price based on age and wheather the person is a student
age = int(input("Enter your age: "))
is_student = input("are you a student ? (yes/no): ").lower()

if age < 5:
    price = "free"
elif age <= 12:
    price = "10$"
elif age <= 17:
    if is_student == "yes":
        price = "12$"
    else:
        price = "15%"
elif age <= 64:
    if is_student == "yes":
        price = "18$"
    else:
        price = "20$"
else:
    price = "25$"

print("The ticket price:", price)    
```
